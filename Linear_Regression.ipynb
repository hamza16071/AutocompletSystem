```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
# Dataset online import

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00477/Real%20estate%20valuation%20data%20set.xlsx"

df = pd.read_excel(url)

df.head()
```

```python
# Dataset Information

df.info()
```

```python
# Dataset Summary

df.describe()
```

```python
# Mising Value Check

df.isnull().sum()
```

```python
# Duplicate Rows

df.duplicated().sum()
```

```python
# Check Column

df.columns
```

```python
# X and Y Define

X = df[["X2 house age"]]

y = df["Y house price of unit area"]
```

```python
# Relationship Visualize

plt.figure(figsize=(8, 5))

plt.scatter(X, y)

plt.xlabel("House Age")
plt.ylabel("House Price per Unit Area")
plt.title("House Age vs House Price")

plt.show()
```

```python
# Train/Test Split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
# Defin or Check Shapes

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
# Create Model

model = LinearRegression()
```

```python
# Model Train

model.fit(X_train, y_train) # .fit (Training data se model ke parameters learn karo.)
```

```python
# Learned parameters

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
# Test on Prediction Data

y_pred = model.predict(X_test)
```

```python
# Actual vs Predicted

comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

comparison.head(10)
```

```python
# Calculate MAE

mae = mean_absolute_error(y_test, y_pred)

print("MAE:", mae)
```

```python
# Calculate MSE

mse = mean_squared_error(y_test, y_pred)

print("MSE:", mse)
```

```python
# Calculate RMSE

rmse = np.sqrt(mse)

print("RMSE:", rmse)
```

```python
# R² model ke explained variance relative

r2 = r2_score(y_test, y_pred)

print("R²:", r2)
```

```python
# Complete evaluation
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
# Actual vs Predicted graph

plt.figure(figsize=(8, 5))

plt.scatter(y_test, y_pred)

plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Actual vs Predicted")

plt.show()
```

```python
# Regression line visualize

plt.figure(figsize=(8, 5))

plt.scatter(X_test, y_test, label="Actual")

plt.plot(
    X_test,
    y_pred,
    label="Regression Line"
)

plt.xlabel("House Age")
plt.ylabel("House Price")
plt.title("Linear Regression")

plt.legend()
plt.show()
```
